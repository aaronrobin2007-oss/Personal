```python
import numpy as np
import matplotlib.pyplot as plt
z=3-4j
r=np.abs(z)
theta_rad=np.angle(z)
theta_deg=np.degrees(theta_rad)
z_cartesian=r*np.exp(1j*theta_rad)
print(f'z={z}, conjugate ={np.conj(z)}')
print(f'polar: r={r:.2f}, theta= {theta_rad:.2f}rad ({theta_deg:.2f}deg)')
plt.axhline(0,color='black',lw=0.8)
plt.axvline(0,color='black',lw=0.8)
plt.plot([0,z.real],[0,z.imag],label='z')
plt.plot([0,z.real],[0,-z.imag],label='z*')
plt.title('argand plane')
plt.show()
print('\n')
plt.figure(figsize=(6,6))
plt.polar([0,theta_rad],[0,r],label='z')
plt.polar([0,-theta_rad],[0,r],label='z*')
plt.title('polar plot', pad=30)
plt.show()
```

```python
import numpy as np
import matplotlib
z=-8+11j
n=3
r=np.abs(z)
theta=np.angle(z)
r_root=r**(1/n)
roots=[]
for k in range(n):
  theta_k= (theta+2*np.pi*k)/n
  z_k= r_root * np.exp(1j*theta_k)
  roots.append(z_k)
  print(f'root,{k}, polar, {r_root:.2f}, {theta_k:.2f}, complex,{z_k:.2f}')
roots_array=np.array(roots)
plt.axhline(0,color='black',lw=0.8)
plt.axvline(0,color='black',lw=0.8)
plt.plot(roots_array.real,roots_array.imag, 'o')
plt.show()
```

```python
import numpy as np
from numpy.polynomial.legendre import leggauss
def f(x):
  return x**2+2*x+1
n=3
x,w=leggauss(n)
a=int(input('enter the lower limit'))
b=int(input('enter the upper limit'))
integral=0
for i in range(n):
  X=((b-a)*x[i]+(a+b))/2
  integral+=f(X)*w[i]
final_ans=(b-a)/2 * integral
print(final_ans)
```

```python
import numpy as np
from numpy.polynomial.laguerre import laggauss
def f(x):
  return x**2+2*x+1
n=3
x,w=laggauss(n)
integral=0
for i in range(n):
  integral+=w[i]*f(x[i])*np.exp(x[i])
print(integral)
```

```python
import numpy as np
from numpy.polynomial.hermite import hermgauss
def f(x):
  return x**2+2*x+1
n=3
x,w=hermgauss(n)
integral=0
for i in range(n):
  integral+=w[i]*f(x[i])*np.exp(x[i]**2)
print(integral)
```

```python
import numpy as np
def f(x):
  return x**2+2*x+11
def p0(x):
  return 1
def p1(x):
  return x
def p2(x):
  return (3*x**2 -1)/2
n=5
x,w=np.polynomial.legendre.leggauss(n)
a0=0
for i in range(n):
  a0+=w[i]*p0(x[i])*f(x[i])
a0=(1/2)*a0
a1=0
for i in range(n):
  a1+=w[i]*p1(x[i])*f(x[i])
a1=(3/2)*a1
a2=0
for i in range(n):
  a2+=w[i]*p2(x[i])*f(x[i])
a2=(5/2)*a2
print(f'{a0:.2f}p0(x)+{a1:.2f}p1(x)+{a2:.2f}p2(x)')
```

```python
import numpy as np
f=np.array([1,2,3,4])
N=len(f)
f_transformed=[]
for k in range(N):
  f_k=0
  for n in range(N):
    f_k+=f[n]*np.exp(-1j*2*np.pi*k*n)
  f_transformed.append(f_k)
print(np.round(f_transformed,2))
```

```python
import numpy as np
f=np.array([1,2,3,4])
n=np.array([0,1,2,3])
N=len(f)
f_transformed=[]
for k in range(N):
  key_k=np.exp(-1j*2*np.pi*n*k)
  score=np.sum(f*key_k)
  f_transformed.append(score)
print(np.round(f_transformed,2))

```

```python
import numpy as np
import matplotlib.pyplot as plt
l=1.0
n=11
x=np.linspace(0,l,n)
t=np.zeros(n)
t[0]=100.0
t[-1]=0.0
for k in range(500):
  for i in range(1,n-1):
    t[i]=0.5*(t[i-1]+t[i+1])
print(np.round(t,2))
print(np.round(x,2))
plt.plot(x,t,marker='o')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
n=11
l=1.0
x=np.linspace(0,l,n)
h=x[1]-x[0]
y=np.zeros(n)
y[0]=1.0
for step in range(1000):
  for i in range(1,n-1):
    y[i]=(y[i+1]+y[i-1])/(2+h**2)
  y[-1]=(2*y[-2])/(2+h**2)
print(np.round(x,2))
print(np.round(y,2))
plt.plot(x,y,marker='o')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
n=20
v=np.zeros([20,20])
v[0,:]=100.0
for step in range(500):
  for i in range(1,n-1):
    for j in range(1,n-1):
      v[i,j]=0.25*(v[i+1,j]+v[i-1,j]+v[i,j+1]+v[i,j-1])
print("centre potential:",np.round(v[n//2,n//2],2))
plt.imshow(v)
plt.show()
```
